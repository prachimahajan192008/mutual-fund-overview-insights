```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scoring_df[['returns_1yr_norm', 'returns_3yr_norm', 'expense_ratio_norm', 'fund_age_norm']] = scaler.fit_transform(
    scoring_df[['returns_1yr', 'returns_3yr', 'expense_ratio', 'fund_age_yr']]
)
```

```python
# Positive 1-year return indicator
scoring_df['positive_1yr_score'] = (scoring_df['returns_1yr'] > 0).astype(float)

# Lower expense ratio = better
scoring_df['low_expense_score'] = 1 - scoring_df['expense_ratio_norm']

# Moderate fund age = better
scoring_df['moderate_age_score'] = 1 - abs(scoring_df['fund_age_norm'] - 0.5) * 2

# Lower risk level = better
risk_scaler = MinMaxScaler()
scoring_df['risk_norm'] = risk_scaler.fit_transform(
    scoring_df[['risk_level']]
)
scoring_df['low_risk_score'] = 1 - scoring_df['risk_norm']

# Final weighted score
scoring_df['score'] = (
    0.40 * scoring_df['returns_3yr_norm'] +
    0.25 * scoring_df['returns_1yr_norm'] +
    0.20 * scoring_df['low_expense_score'] +
    0.10 * scoring_df['moderate_age_score'] +
    0.05 * scoring_df['low_risk_score']
)
```
